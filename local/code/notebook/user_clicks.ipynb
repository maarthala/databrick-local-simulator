```python
from pyspark.sql import SparkSession

spark = SparkSession.builder \
    .appName("Read data using Hive Metastore stored in s3 s3://demo-bucket/clicks/") \
    .enableHiveSupport() \
    .getOrCreate()

print("Hadoop version:", spark.sparkContext._jvm.org.apache.hadoop.util.VersionInfo.getVersion())
```

```python
df = spark.table("clickstream.user_clicks")
df.show(15)
```
